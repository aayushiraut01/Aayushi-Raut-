```python
class person:
    company="TCS"
    def __init__(self,name,proffession,dob,education):
        self.name=name
        self.proffession=proffession
        self.dob=dob
        self.education=education
    def age (self, current_year):
        return current_year-self.dob
    def new_company(self,new_company):
        self.company=new_company
        return new_company
obj=person("Vikas","Trainer",1997,"B.tech")
```

```python
obj.age(2026)
```
