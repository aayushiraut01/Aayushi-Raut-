```python
year=2004
if year %4==0 and year %100!=0:
    print("leap year")
elif year %400==0:
    print("leap year")
else:
    print("not a leap year")
```

```python
numbers = [12, 7, 34, 21, 5, 10, 8, 3, 19, 2]
list_even=[]
list_odd=[]
for i in numbers:
    if i %2==0:
        list_even.append(i)
    else:
        list_odd.append(i)
        
print(list_even)
print(list_odd)
```
