```python

numbers = [8, 15, 3, 21, 12, 18]
largest=numbers[0]
second=numbers[1]
if largest < second:
    largest, second = second , largest

for i in numbers:
    if i > largest:
        second=largest
        largest=i
    elif i > second:
        second= i
print(second)
```
