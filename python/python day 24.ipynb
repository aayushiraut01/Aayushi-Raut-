```python
def merge_list(list1,list2):
    result=[]
    for i in list1:
        if i %2!=0:
            result.append(i)
    for i in list2:
        if i %2==0:
            result.append(i)
    return result
            
list1 = [10, 20, 25, 30, 35]
list2 = [40, 45, 60, 75, 90]
print(merge_list(list1,list2))
```

```python
def exponent(base, exp):
    num = exp
    result = 1
    # Repeat multiplication 'exp' times
    while num > 0:
        result = result * base
        num = num - 1
    print(base, "raises to the power of", exp, "is:", result)

exponent(2, 5)
exponent(5, 4)
```

```python
def exponent(base,expo):
    num=expo
    result=1
    while num>0:
        result=result*base
        num=num-1
    return result
exponent(2, 5)
exponent(5, 4)
print(exponent(2,5))
print(exponent(5,4))
    
```
