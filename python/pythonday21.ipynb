```python
numbers = [3, 4, 3, 2, 4, 3, 5]
frequency={}
for i in numbers:
    if i in frequency:
        frequency[i]=frequency[i]+1
    else:
        frequency[i]=1
higest_frequency=0
most_frequent=None
for number, count in frequency.items():
    if count > higest_frequency:
        higest_frequency=count
        most_frequent=number
print("higest_frequency=",higest_frequency)
print("most_frequent=",most_frequent)
        
```

```python
numbers = [2, 4, 2, 7, 4, 9, 7]

seen = set()
unique = []

for i in numbers:
    if i not in seen:
        seen.add(i)
        unique.append(i)

print(unique)
```
