```python
numbers = [12, 5, 7, 12, 8, 5, 19, 7, 12, 20]

seen=set()
largest=float("-inf")
second=float("-inf")
for i in numbers:
    if i in seen:
        continue

    seen.add(i)
    if i > largest:
        second=largest
        largest=i
    elif i > second:
        second=i
print(second)
```

```python
numbers = [10, 20, 5, 20, 8, 15, 10, 18]

def second_largest(numbers):
    seen = set()
    largest = float('-inf')
    second = float('-inf')

    for i in numbers:
        if i in seen:
            continue

        seen.add(i)

        if i > largest:
            second = largest
            largest = i

        elif i > second:
            second = i

    return second

print(second_largest(numbers))
```

```python
numbers = [2, 5, 2, 8, 5, 9, 2, 8, 10]
def duplicate(numbers):
    seen=set()
    duplicates=set()
    for i in numbers:
        if i in seen:
            duplicates.add(i)
        else:
            seen.add(i)
    return duplicates
    
print(duplicate(numbers))
```
