```python
numbers = [23, 7, 45, 12, 89, 34, 56]

try:
    largest = numbers[0]
    second = numbers[1]

    if largest < second:
        largest, second = second, largest

    for i in numbers[2:]:
        if i > largest:
            second = largest
            largest = i
        elif i > second:
            second = i

    print(second)

except Exception as e:
    print(e)
    
```

```python
numbers = [2, 4, 2, 7, 4, 2, 9, 4]

frequency = {}

# Step 1: Count frequency
for i in numbers:
    if i in frequency:
        frequency[i] += 1
    else:
        frequency[i] = 1

# Step 2: Find most frequent number
highest = 0
most_frequent = 0

for number, count in frequency.items():
    if count > highest:
        highest = count
        most_frequent = number

print("Most frequent =", most_frequent)
print("Frequency =", highest)
```
