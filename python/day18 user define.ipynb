```python
def largest(a, b):
    if a > b:
        return a
    else:
        return b

print(largest(15, 10))
```

```python
def largest(a, b, c):
    if a >= b and a >= c:
        return a
    elif b >= a and b >= c:
        return b
    else:
        return c

print(largest(10, 25, 15))
```

```python
def factorial(n):
    fact = 1

    for i in range(1, n + 1):
        fact = fact * i

    return fact

print(factorial(5))
```

```python
def sum_numbers(n):
    total = 0

    for i in range(1, n + 1):
        total = total + i

    return total

print(sum_numbers(5))
```

```python
def count_even(n):
    count = 0

    for i in range(1, n + 1):
        if i % 2 == 0:
            count = count + 1

    return count

print(count_even(10))
```

```python
def table(n):
    for i in range(1, 11):
        print(n, "*", i, "=", n * i)

table(5)
```

```python
def count_vowels(word):
    count = 0

    for i in word:
        if i in "aeiou":
            count = count + 1

    return count

print(count_vowels("banana"))
```

```python
def reverse_string(word):
    reverse = ""

    for i in word:
        reverse = i + reverse

    return reverse

print(reverse_string("python"))
```

```python
def find_largest(numbers):
    largest = numbers[0]

    for i in numbers:
        if i > largest:
            largest = i

    return largest

numbers = [10, 25, 7, 40, 15]
print(find_largest(numbers))
```

```python
def find_smallest(numbers):
    smallest = numbers[0]

    for i in numbers:
        if i < smallest:
            smallest = i

    return smallest

numbers = [10, 25, 7, 40, 15]
print(find_smallest(numbers))
```

```python
def count_even_odd(numbers):
    even = 0
    odd = 0

    for i in numbers:
        if i % 2 == 0:
            even = even + 1
        else:
            odd = odd + 1

    print("Even =", even)
    print("Odd =", odd)

numbers = [2, 5, 8, 11, 14, 17]
count_even_odd(numbers)
```

```python
def find_duplicates(numbers):
    seen = set()
    duplicates = set()

    for i in numbers:
        if i in seen:
            duplicates.add(i)
        else:
            seen.add(i)

    return duplicates

numbers = [2, 4, 2, 7, 4, 2, 9, 4]
print(find_duplicates(numbers))
```
