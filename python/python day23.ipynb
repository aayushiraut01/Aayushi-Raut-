```python
def palindrome(numbers):
    numbers=str(numbers)
    if numbers == numbers[::-1]:
        return True
    else:
        return False
number1 = 121
number2 = 125
print(palindrome(number1))
print(palindrome(number2))
```

```python
def check_palindrome(number):
    number = str(number)

    if number == number[::-1]:
        return True
    else:
        return False


number1 = 121
number2 = 125

print(check_palindrome(number1))
print(check_palindrome(number2))
```

```python
def check_palindrome(number):
    print("original number", number)
    # Convert to string to reverse easily
    original_str = str(number)
    reversed_str = original_str[::-1]
    
    if original_str == reversed_str:
        print("Yes. given number is palindrome number")
    else:
        print("No. given number is not palindrome number")

check_palindrome(121)
check_palindrome(125)
```
