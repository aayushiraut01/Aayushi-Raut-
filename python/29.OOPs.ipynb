```python
class vehicle:
    def info(self):
        print("great mileage")


class car(vehicle):
    def wheels(self):
        print("has 4 wheels")


class bike(vehicle):
    def electric(self):
        print("its electric")


class electric(car, bike):
    def both(self):
        print("both are electric")
```

```python
obj=electric()
```

```python
obj.both()
```
