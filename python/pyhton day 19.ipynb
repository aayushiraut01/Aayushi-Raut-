```python
numbers = [2, 4, 2, 7, 4, 2, 9, 4]
def dup(numbers):
    count=0
    seen=set()
    for i in numbers:
        if i in seen:
            count=count+1
        else:
            seen.add(i)
    return count
print(dup(numbers))
        
```

```python
def frequency(numbers):
    freq = {}

    for i in numbers:
        if i in freq:
            freq[i] = freq[i] + 1
        else:
            freq[i] = 1

    return freq

numbers = [2, 4, 2, 7, 4, 2, 9, 4]
print(frequency(numbers))
```

```python
def prime_numb(number):
    if number < 2:
        return False 
    for i in range (2,number):
        if number %i==0:
            return False 
    return True 
print(prime_numb(19))
```

```python
def prime_number(number):
    if number <2:
        return False 
    for i in range (2,number):
        if number %i==0:
            return False
    return True
print(prime_number(71))
```
