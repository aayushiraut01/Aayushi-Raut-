```python
class Vehicle:
    def start(self):
        print("vehicle started")
        

class car(Vehicle):
    def drive(self, petrol, disel, electric):
        self.petrol=petrol
        self.disel=disel
        self.eletric=electric
        print("driving a car")

class bike(Vehicle):
    def ride(self, petrol, electric):
        self.petrol=petrol
        self.electric=electric
        print("riding a bike")
```

```python
obj.start()
```
